```python
# Change the element based on the index

A=["disco",10,1.2]
print('Before change:',A)
A[0]='hard rock'
print('After change:',A)
```

```python
# Delete the element based on the index

print('Before change:', A)
del(A[0])
print('After change', A)
```

```python
# Copy (Copy reference) the list A

A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
```

```python
# Examine the copy by reference

print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```

```python
# Clone (clone by value) the list A

B = A[:]
B
```

```python
print('B[0]:', B[0])
A[0] = "hard rock"
print('B[0]:', B[0])
```
